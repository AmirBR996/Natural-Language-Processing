This cell imports the `nltk` library and downloads the 'wordnet' corpus, which is essential for lemmatization.

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
```

This cell defines a list of words that will be used as input for the lemmatization examples.

```python
words = ['eating' , 'running' , 'eaten' , 'run' , 'visiting' , 'visited' , 'eated']
```

This cell initializes an instance of the `WordNetLemmatizer`. This object will be used to call the `lemmatize` method.

```python
lemmatizer = WordNetLemmatizer()
```

This cell provides a comment block explaining the part-of-speech (pos) tags used by the `lemmatize` function in NLTK's WordNetLemmatizer. 'n' is for Noun, 'v' for Verb, 'a' for Adjective, and 'r' for Adverb.

```python
# pos
# Noun -> n
# Verb -> v
# Adjective -> a
# adverb -> r
```

This cell iterates through the `words` list and prints the lemmatized form of each word, explicitly specifying `pos="v"` to treat them as verbs. This means the lemmatizer will try to find the base form of the verb.

```python
for word in words:
  print(lemmatizer.lemmatize(word , pos="v"))
```

This cell iterates through the `words` list again, but this time it prints the lemmatized form of each word while specifying `pos="a"` to treat them as adjectives. Notice how the output might differ from verb lemmatization.

```python
for word in words:
  print(lemmatizer.lemmatize(word , pos="a"))
```
