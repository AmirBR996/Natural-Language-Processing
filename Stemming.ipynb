This cell initializes a list of words that will be used to demonstrate various stemming algorithms. These words are chosen to showcase how different stemmers handle various suffixes and irregular forms.

```python
words = ['eating' , 'running' , 'eaten' , 'run' , 'visiting' , 'visited' , 'eated']
```

This block imports the `PorterStemmer` from NLTK. The Porter Stemmer is a classic stemming algorithm that uses a set of rules to remove common suffixes from words, aiming to reduce words to their base or root form (stem).

```python
# porterstemmer

from nltk.stem import PorterStemmer
portstemmer = PorterStemmer()
```

This cell iterates through the `words` list and applies the `portstemmer` (Porter Stemmer) to each word. The output demonstrates how the Porter Stemmer reduces words like 'eating', 'running', 'visiting' to their stems 'eat', 'run', 'visit', respectively, but also shows some limitations like 'eaten' and 'eated' both becoming 'eat'.

```python
for word in words:
  print(portstemmer.stem(word))
```

This cell highlights a potential disadvantage of the Porter Stemmer. When stemming 'congratulations', it reduces it to 'congratul', which is not a recognized word or a clear stem. This illustrates that rule-based stemmers can sometimes produce non-dictionary words or over-stem.

```python
# disadvantages
portstemmer.stem("congratulations")
```

This block introduces the `RegexpStemmer` from NLTK. Unlike the Porter Stemmer, which is rule-based, the Regexp Stemmer uses regular expressions to define which suffixes to remove. Here, it's initialized to remove 'ing', 's', 'e', 'able', 'en', and 'ed' suffixes, provided the word has a minimum length of 4 characters after stemming.

```python
# regexpstemmer
from nltk.stem import RegexpStemmer

reg_stemmer = RegexpStemmer(regexp= 'ing$|s$|e$|able$|en$|ed$', min=4)
```

This cell applies the `reg_stemmer` (Regexp Stemmer) to the `words` list. Notice how it produces different results compared to the Porter Stemmer, especially for 'running' becoming 'runn' because 'ing' was removed, and 'eaten' becoming 'eat' due to the 'en' removal rule.

```python
for word in words:
  print(reg_stemmer.stem(word))
```

This block imports the `SnowballStemmer` from NLTK. The Snowball Stemmer (also known as Porter2) is an improvement over the original Porter Stemmer. It supports multiple languages and is generally more aggressive and produces better results by addressing some of the issues found in the Porter Stemmer.

```python
# snowball stemmer
from nltk.stem import SnowballStemmer
snowball = SnowballStemmer('english')
```

This cell demonstrates the `snowball` stemmer on the `words` list. You can observe how it handles words more accurately than the Porter Stemmer, for example, 'eating' and 'eaten' both stem to 'eat', and 'running' and 'run' both stem to 'run', which is often the desired behavior for a stemmer.

```python
for word in words:
  print(snowball.stem(word))
```
