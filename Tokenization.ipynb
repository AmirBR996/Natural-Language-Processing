This cell is responsible for setting up the NLTK (Natural Language Toolkit) library, which is fundamental for many NLP operations. It imports NLTK and then downloads two crucial resources: `punkt` (a pre-trained tokenizer for English, used for sentence and word tokenization) and `punkt_tab` (another tokenizer, often used for more specialized cases or as a dependency).

```python
# !pip install nltk
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
```

This cell defines a multi-line string variable named `corpus`. This string contains information about Nepal and will serve as our primary text data for various Natural Language Processing (NLP) tasks in this notebook.

```python
corpus = """Nepal, officially the Federal Democratic Republic of Nepal, is a landlocked country in South Asia. It is mainly situated in the Himalayas, but also includes parts of the Indo-Gangetic Plain. It borders the Tibet Autonomous Region of China to the north, and India to the south, east, and west, while it is narrowly separated from Bangladesh by the Siliguri Corridor, and from Bhutan by the Indian state of Sikkim. Nepal has a diverse geography, including fertile plains, subalpine forested hills, and eight of the world's ten highest mountains, including Mount Everest, the highest point above mean sea level on Earth. Kathmandu is the nation's capital and its largest city. Nepal is a multi-ethnic, multi-lingual, multi-religious, and multi-cultural sovereign state, with Nepali as the official language."""
```

This cell simply prints the entire content of the `corpus` variable to the output. This allows for a quick verification of the text data we are working with.

```python
print(corpus)
```

This cell imports the `sent_tokenize` function from `nltk.tokenize`. It then uses this function to split the `corpus` into individual sentences. The result is a list of strings, where each string represents a detected sentence, and is stored in the `document` variable.

```python
from nltk.tokenize import sent_tokenize

document = sent_tokenize(corpus)
```

This cell iterates through the `document` list (which contains the individual sentences generated by `sent_tokenize`). For each sentence, it prints the sentence on a new line, providing a clear view of how the original `corpus` was segmented into sentences.

```python
for sentences in document :
  print(sentences)
```

This cell imports the `word_tokenize` function from `nltk.tokenize`. It then applies this function to the `corpus` to break it down into individual words and punctuation marks (tokens). The resulting list of tokens is stored in the `words` variable.

```python
from nltk.tokenize import word_tokenize

words = word_tokenize(corpus)
```

This cell iterates through the `words` list, which contains the word and punctuation tokens generated by `word_tokenize`. It prints each token on a new line, allowing us to inspect the granular breakdown of the text into its constituent parts.

```python
for words in words:
  print(words)
```

This cell imports `wordpunct_tokenize` from `nltk.tokenize` and applies it to the `corpus`. `wordpunct_tokenize` is another word tokenizer that tends to split words and punctuation more aggressively than `word_tokenize`. For example, it might separate a word from an apostrophe more readily. The output is a list of these more finely-grained tokens.

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

This cell introduces the `TreebankWordDetokenizer` from NLTK. This class is primarily designed for **detokenization**, which is the process of reconstructing a string from a list of tokens (e.g., combining `['don', "'t"]` into `"don't"`).

However, in this specific code, `tokenizer.tokenize(corpus)` is being called directly on the raw `corpus` string. The `tokenize` method of `TreebankWordDetokenizer` is not intended for standard word tokenization of raw text. When given a raw string, it processes each character as a potential token and then attempts to 'detokenize' them, leading to an output where each character of the original text is separated by spaces. This is a common point of confusion as its `tokenize` method here doesn't perform traditional tokenization of a string into words.

```python
from nltk.tokenize import TreebankWordDetokenizer

tokenizer = TreebankWordDetokenizer()

tokenizer.tokenize(corpus)
```
