```python
paragraph = """Quantum computing is an emerging technology that harnesses quantum mechanics—specifically superposition and entanglement—to process information in fundamentally new ways. Unlike classical computers that use binary bits (0 or 1), quantum machines utilize quantum bits, or qubits, which can exist in multiple states simultaneously, allowing them to tackle complex optimization, simulation, and cryptographic tasks exponentially faster than traditional supercomputers. While still in experimental stages, these systems promise breakthroughs in fields like medicine, AI, and material science by solving problems deemed impossible for conventional technology."""
```

```python
from nltk.stem import WordNetLemmatizer
from nltk.corpus import stopwords
import nltk
nltk.download("stopwords")
```

```python
lemmatize = WordNetLemmatizer()
```

```python
nltk.download("wordnet")
```

```python
from nltk.tokenize import sent_tokenize

sentences = sent_tokenize(paragraph)
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [
        lemmatize.lemmatize(word.lower() , "v")
        for word in words
        if word.lower() not in set(stopwords.words("english"))
    ]

    sentences[i] = " ".join(words)
```
